```python
from pydantic import BaseModel, Field
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    temperature=0.7,
)


class Country(BaseModel):
    name: str = Field(..., description="The name of the country")
    language: str = Field(..., description="The language of the country")
    capital: str = Field(..., description="The capital city of the country")

```

```python
structured_llm = llm.with_structured_output(Country, method="json_schema")

result = structured_llm.invoke("Tell me about Pakistan")
print(result)
```

```python
from typing import Optional
from typing_extensions import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0.7,
)

class Joke(TypedDict):
    """Joke to tell user.

    Attributes:
        setup: The setup of the joke
        punchline: The punchline of the joke
        rating: How funny the joke is on a scale of 1-10
    """
    setup: str
    punchline: str
    rating: Optional[int]

structured_llm = llm.with_structured_output(Joke)
structured_result = structured_llm.invoke("Tell me a joke about programming")
print(structured_result)
```

```python
json_schema = {
    "title": "Joke",
    "description": "Joke to tell user.",
    "type": "object",
    "properties": {
        "setup": {
            "title": "Setup",
            "description": "The setup of the joke",
            "type": "string",
        },
        "punchline": {
            "title": "Punchline",
            "description": "The punchline of the joke",
            "type": "string",
        },
        "rating": {
            "title": "Rating",
            "description": "How funny the joke is on a scale of 1-10",
            "default": None,
            "type": ["integer", "null"],
        },
    },
}

structured_llm = llm.with_structured_output(schema=json_schema)
structured_result = structured_llm.invoke("Tell me a joke about programming")
print(structured_result)
```
